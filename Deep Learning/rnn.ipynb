# RNN (Recurrent Neural Network)

### Airline Passengers (CSV)

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense
```

```python
data = pd.read_csv(r"C:\Users\Karan\Downloads\airline-passengers.csv")
values = data['Passengers'].values.reshape(-1,1)
scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(values)
```

```python
x = []
y = []
for i in range(12, len(scaled_data)):
    x.append(scaled_data[i-12:i ,0])
    y.append(scaled_data[i ,0])

x = np.array(x)
y = np.array(y)
```

```python
x.shape
```

```python
split = int(len(x) * 0.8)
x_train = x[:split]
x_test = x[split:]

y_train = y[:split]
y_test = y[split:]
```

```python
model = Sequential([
                    SimpleRNN(50, activation='tanh', input_shape=(12,1)),
                    Dense(1)
                    ])
model.compile(optimizer='adam', loss='mse', metrics=['mae'])
model.fit(x_train, y_train, epochs=100, batch_size=16, validation_data=(x_test, y_test), verbose=1)
predictions = model.predict(x_test)
```

```python
predictions = scaler.inverse_transform(predictions)
actual = scaler.inverse_transform(y_test.reshape(-1, 1))
mae = mean_absolute_error(actual, predictions)
print("Mean Absolute Error:", mae)
```

### Next Number Prediction

```python
data = np.array([10,20,30,40,50,60,70,80])
data = data.reshape(-1,1)
x = []
y = []

scaler = MinMaxScaler()
scaled = scaler.fit_transform(data).flatten()

time_steps = 3
for i in range(len(scaled) - time_steps): #8-3=5
    x.append(scaled[i:i+time_steps])
    y.append(scaled[i+time_steps])

print(x)
print(y)
```

```python
x = np.array(x)
y = np.array(y)

# Reshape for RNN
# (samples, time_steps, features)

x = x.reshape(x.shape[0], 3, 1)
print('X Shape:', x.shape)
print('Y Shape:', y.shape)

model = Sequential([SimpleRNN(32, activation='tanh', input_shape=(3,1)),
                    Dense(1)])

model.compile(optimizer='adam', loss='mse')
model.fit(x,y,epochs=1000, verbose=1)
```

```python
test_input = np.array([70,80,90])
test_input = test_input.reshape(1,3,1)
prediction = model.predict(test_input, verbose=0)
print('Next Number Predicted:', scaler.inverse_transform(prediction)[0][0])
```

### Alice in the Wonderland

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.layers import Embedding, Dropout

url = "https://www.gutenberg.org/files/11/11-0.txt"
path = tf.keras.utils.get_file("alice.txt",url)
with open(path, "r", encoding="utf-8") as file:
    text = file.read()
```

```python
print('Total Characters:', len(text))
tokenizer = Tokenizer(oov_token='<OOV>')
tokenizer.fit_on_texts([text])
total_words = len(tokenizer.word_index) + 1
print('Total Vocabulary:', total_words)

input_sequences = []
for line in text.split('\n'):
    token_list = tokenizer.texts_to_sequences([line])[0]
    for i in range(1, len(token_list) + 1):
        ngram_sequence = token_list[:i]
        if len(ngram_sequence) > 1:
            input_sequences.append(ngram_sequence)

print('Total sequences:', len(input_sequences))
input_sequences
```

```python
max_sequence_len = max(len(sequence) for sequence in input_sequences)
input_sequences = np.array(pad_sequences(input_sequences, maxlen=max_sequence_len,
                                         padding='pre'))
print('Sequence Shape:', input_sequences.shape)
input_sequences
```

```python
x=input_sequences[:,:-1]
y=input_sequences[:,-1]
print(x)
print(y)
```

```python
from tensorflow.keras.layers import Embedding,Dropout,LSTM
model = Sequential([
    Embedding(input_dim=total_words,output_dim=100,mask_zero=True),
    LSTM(128, activation="tanh"),
    Dropout(0.2),
    Dense(total_words, activation="softmax")
])
model.compile(optimizer="adam",loss="sparse_categorical_crossentropy",metrics=["accuracy"])
model.fit(x,y,epochs=20,batch_size=128,validation_split=0.1)
```

```python
def predict_next_word(seed_text):
    token_list = tokenizer.texts_to_sequences([seed_text])[0]
    if len(token_list) == 0:
        return 'Uknown input'
    token_list = pad_sequences([token_list], maxlen=max_sequence_len-1,
                               padding='pre', truncating='pre')
    predicted = model.predict(token_list, verbose =0)
    predicted_word_index = np.argmax(predicted)
    for word, index in tokenizer.word_index.items():
        if index==predicted_word_index:
            return word
    return 'Unknown'
```

```python
while True:
    user_in=input("Enter text: ")
    if user_in=="exit":
        print('Exited')
        break
    res=predict_next_word(user_in)
    print(user_in,":", res)
```
